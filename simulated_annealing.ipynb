```python

import random
import math

# Cost function
def cost(x):
    return x * x

# Generate random neighbour
def neighbour(current):
    return current + random.choice([-1, 1])

# Simulated Annealing
def simulated_annealing(current):
    T = 100

    while T > 0.1:
        next_state = neighbour(current)

        deltaE = cost(current) - cost(next_state)

        if deltaE > 0:
            current = next_state
        else:
            p = math.exp(deltaE / T)

            if random.random() < p:
                current = next_state

        # Decrease temperature
        T = T * 0.95

    return current


# Dynamic input
initial = int(input("Enter initial state: "))

result = simulated_annealing(initial)

print("Final state:", result)
print("Minimum cost:", cost(result))
```
