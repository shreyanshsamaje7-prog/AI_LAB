```python
# Dynamic Forward Chaining Algorithm

facts = set()

# Get initial facts
n = int(input("Enter number of initial facts: "))

for i in range(n):
    fact = input("Enter fact: ").strip()
    facts.add(fact)

# Get rules
rules = []
r = int(input("\nEnter number of rules: "))

for i in range(r):
    print(f"\nRule {i+1}")

    conditions = input("Enter conditions (comma separated): ")
    conditions = set(x.strip() for x in conditions.split(","))

    conclusion = input("Enter conclusion: ").strip()

    rules.append((conditions, conclusion))

# Get goal
goal = input("\nEnter goal: ").strip()

# Forward Chaining
while True:
    new = False

    for conditions, conclusion in rules:

        if conditions.issubset(facts):
            if conclusion not in facts:
                facts.add(conclusion)

                print("New fact generated:", conclusion)
                new = True

    if goal in facts:
        print("\nGoal reached:", goal)
        break

    if not new:
        print("\nGoal cannot be reached!")
        break

print("\nFinal Knowledge Base:", facts)
```
