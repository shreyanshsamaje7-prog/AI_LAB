```python
def unify(x, y, subs):
    if subs is None:
        return None

    x = substitute(x, subs)
    y = substitute(y, subs)

    if x == y:
        return subs

    if is_variable(x):
        return unify_var(x, y, subs)

    if is_variable(y):
        return unify_var(y, x, subs)

    if isinstance(x, tuple) and isinstance(y, tuple):

        if x[0] != y[0] or len(x) != len(y):
            return None

        for a, b in zip(x[1:], y[1:]):
            subs = unify(a, b, subs)

            if subs is None:
                return None

        return subs

    return None


def is_variable(x):
    return isinstance(x, str) and x.startswith("?")


def substitute(x, subs):
    while is_variable(x) and x in subs:
        x = subs[x]

    if isinstance(x, tuple):
        return tuple([x[0]] + [
            substitute(a, subs) for a in x[1:]
        ])

    return x


def occurs_check(var, expr, subs):
    expr = substitute(expr, subs)

    if var == expr:
        return True

    if isinstance(expr, tuple):
        return any(
            occurs_check(var, a, subs)
            for a in expr[1:]
        )

    return False


def unify_var(var, expr, subs):
    if var in subs:
        return unify(subs[var], expr, subs)

    if occurs_check(var, expr, subs):
        return None

    subs[var] = expr
    return subs


# Read expressions dynamically
def read_expression():
    name = input("Enter predicate name: ").strip()

    n = int(input("Enter number of arguments: "))

    args = []

    for i in range(n):
        value = input(f"Enter argument {i+1} (use ? for variable): ").strip()
        args.append(value)

    return tuple([name] + args)


# Main program
print("FIRST ORDER LOGIC - UNIFICATION")

print("\nEnter First Expression:")
e1 = read_expression()

print("\nEnter Second Expression:")
e2 = read_expression()

result = unify(e1, e2, {})

if result is not None:
    print("\nUnification Successful!")
    print("MGU:", result)
else:
    print("\nUnification Failed!")
```
